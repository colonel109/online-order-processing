```python
import pandas as pd
from pathlib import Path
import duckdb
```

```python
control_file = Path(r"E:\online-order-processing\shopee_control_file.xlsx")
```

```python
combo_df = pd.read_excel(control_file, sheet_name="combo")
promotions_df = pd.read_excel(control_file, sheet_name="promotions")

duckdb.sql("""
    SELECT
        combo_key,
        combo_name,
        product_code,
        product_name,
        product_price,
        quantity,
        -- Tổng giá trị của từng dòng sản phẩm
        (product_price * quantity) AS line_item_value,
        -- Tổng giá trị của TOÀN BỘ combo (cộng dồn các dòng có cùng combo_key)
        SUM(product_price * quantity) OVER(PARTITION BY combo_key) AS total_combo_value
    FROM promotions_df
    LEFT JOIN combo_df USING (combo_key)
""").to_df()
```
